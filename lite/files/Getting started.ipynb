# C++ in your browser

Choose **Run → Run All Cells** to try these examples. The C++23 kernel
runs locally in a browser worker, using xeus-cpp and WebAssembly.
The first start downloads its runtime; no remote server executes your code.

For a separate C23 kernel, open **C examples.ipynb**.


```cpp
#include <iostream>
#include <numeric>
#include <vector>

std::vector<int> values = {6, 12, 24};
std::cout << "C++ total: "
          << std::accumulate(values.begin(), values.end(), 0) << "\n";
```

## Reuse definitions across cells

Notebook cells share kernel state. Change the input and rerun the next cell.
Restart the kernel to clear its definitions and start over.


```cpp
int square(int value) { return value * value; }
```

```cpp
std::cout << "square(7): " << square(7) << "\n";
```

## Inspect the generated assembly

Open a launcher with **File → New Launcher**, then select **Open Fortitudo**.
Paste the `square` function into the source editor and choose **Compile**.
Try changing the optimization level or target to compare the assembly.

The explorer uses its own LLVM compiler worker. Notebook execution uses
a separate interpreter, so their state and compiler versions are independent.
Native processes, arbitrary native libraries, and native platform APIs are
not available in this browser environment.
