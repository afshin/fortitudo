# C in your browser

Choose **Run → Run All Cells** to run these examples with the C23 kernel.
The xeus-cpp interpreter runs locally in WebAssembly.


```c
#include <stdio.h>

int square(int value) { return value * value; }
```

```c
printf("C square(7): %d\n", square(7));
```

## Shared state

Definitions stay available in later cells until you restart the kernel.
This example uses C standard library headers packaged with the runtime.


```c
int values[] = {6, 12, 24};
int total = 0;
for (int i = 0; i < 3; ++i) {
    total += values[i];
}
printf("C total: %d\n", total);
```
